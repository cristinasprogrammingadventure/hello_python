```python
# Crea dues variables numèriques (edats) i indica quina de les dues és més gran.

adrian=17
paula=4
adrian > paula
```

```python

# mostra per pantalla un missatge si el valor està 
# entre 1 i 10, un altre missatge si el valor està entre 11 i 20 i un altre missatge si el valor està per sobre de 20.

adrian=17
paula=4

if adrian <=10 and adrian >0 :
    print("Adrián tiene menos de 11 años")
elif adrian >=11 and adrian < 20 :
    print("Adrián tiene entre 11 y 20 años")
elif adrian >= 20 :
    print("Adrián tiene más de 20 años")
else :
        print("La edat no es válida")
```

```python
if paula <=10 and paula >0 :
    print("paula tiene menos de 11 años")
elif paula >=11 and paula < 20 :
    print("paula tiene entre 11 y 20 años")
elif paula >= 20 :
    print("paula tiene más de 20 años")
else :
        print("La edat no es válida")
```

```python
adrian = 7
paula = 4

if adrian > paula:
    print("Adrián es mayor que Paula")
elif adrian < paula:
    print("Paula es mayor que Adrián")
else:
    print("Ambos niños tienen la misma edad")
```
